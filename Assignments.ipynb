<b>Assignment 1: N-Queens</b>

### Problem statement

Place **n queens** on an **n × n** chessboard so that no two queens share a **row**, a **column**
or a **diagonal**.

### Required representation

Represent a state as a **tuple of length n**, where `state[c]` is the **row** of the queen in
**column c** (rows and columns numbered 0 to n − 1). For example, for n = 4 the state `(1, 3, 0, 2)`
places the queen of column 0 in row 1, the queen of column 1 in row 3, and so on.

Because there is exactly one queen per column, **column conflicts are impossible by
construction**. Only **rows** and **diagonals** need to be checked. Two queens in columns c1 and c2
attack each other diagonally when |state[c1] − state[c2]| = |c1 − c2|.

### Choice of algorithms

Choose **two** algorithms from the list below.

| Option | Algorithm | Formulation |
|---|---|---|
| **A** | DFS with backtracking | state = partial placement, one queen per column (left to right); prune a branch as soon as the newly placed queen conflicts with an earlier one |
| **B** | Hill Climbing | state = full assignment (n queens); neighbour = move one queen to a different row **within its own column**; cost = number of attacking pairs; cost 0 = solved |
| **C** | Simulated Annealing | same formulation as B |
| **D** | Genetic Algorithm | same formulation as B |

**Constraint:** your two algorithms must **not both** come from {B, C, D}. **One of the two must be
A**, so that your comparison contrasts a **systematic search** with a **local search**. Valid
choices are therefore A+B, A+C or A+D.

**Not allowed:** **BFS, IDS, UCS, Greedy Best-First and A\* must NOT be used in this assignment.**
(Deliverable f asks you to explain why.)
